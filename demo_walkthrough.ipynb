## 📌 Milestone M1: Schemas & JSON Utilities
In this milestone, we implemented:
1. Strict data schemas (`ProfileJSON`, `PostingJSON`, `MatchResult`, `Draft`) using Pydantic v2 in `jsb/schemas.py`.
2. Resilient JSON extraction in `jsb/json_utils.py` that strips markdown code fences, handles conversational wrapper text, and retries up to 2 times on malformed outputs.


### 1. Data Schemas (`jsb/schemas.py`)
Let us test instantiating the core schemas with sample data.

````python
from jsb.schemas import ProfileJSON, PostingJSON, MatchResult, Draft, RoleItem, BudgetInfo

profile = ProfileJSON(
    name='Alex Mercer',
    headline='Senior AI Engineer',
    skills=['python', 'langchain', 'faiss', 'fastapi', 'docker', 'sql', 'git'],
    years_experience=3.5,
    roles=[RoleItem(title='Senior AI Developer', company='Innovatech', years=2.0, highlights=['Built RAG'])],
    education=['B.S. Computer Science']
)
print(f'✅ ProfileJSON: {profile.name} ({profile.years_experience} yrs experience, {len(profile.skills)} skills)')

posting = PostingJSON(
    title='Machine Learning Engineer',
    company_or_client='Visionary Labs',
    type='job',
    must_have=['python', 'pytorch'],
    nice_to_have=['docker']
)
print(f'✅ PostingJSON: {posting.title} at {posting.company_or_client} ({posting.type})')

match = MatchResult(
    score=85.0,
    must_coverage=1.0,
    nice_coverage=0.8,
    verdict='APPLY',
    reasons=['Full must-have coverage', 'Solid background'],
    advice='Proceed with application.'
)
print(f'✅ MatchResult: Score {match.score} -> Verdict: {match.verdict}')

draft = Draft(kind='cover_letter', subject='Application', body='Dear Team...', claims_ok=True)
print(f'✅ Draft: {draft.kind} (Claims Verified: {draft.claims_ok})')
````

### 2. Code Fence Stripping & Parsing (`strip_fences` & `parse_raw_json`)
LLMs frequently output markdown fences like ` ```json ... ``` ` or conversational chatter before/after the JSON. Our utilities cleanly isolate the JSON.

````python
from jsb.json_utils import strip_fences, parse_raw_json

messy_llm_output = (
    'Sure! Here is the extracted profile:\n'
    '```json\n'
    '{\n'
    '  "name": "Jane Doe",\n'
    '  "skills": ["python", "fastapi"]\n'
    '}\n'
    '```\n'
    'Hope that helps!'
)
print('Raw LLM Output:\n', messy_llm_output)
parsed = parse_raw_json(messy_llm_output)
print('Parsed cleanly into Python dict:\n', parsed)
````

### 3. Validated Schema Extraction & Auto-Retry (`extract_json`)
If an LLM produces broken JSON, `extract_json` asks the LLM to fix its output and validates it against the schema.

````python
from jsb.json_utils import extract_json, safe_extract_json

# 1. Successful extraction
print('1. Testing extract_json on messy output...')
extracted_profile = extract_json(messy_llm_output, ProfileJSON)
print(f'Extracted Profile: {extracted_profile.name}, Skills: {extracted_profile.skills}\n')

# 2. Simulated self-repair on broken JSON
print('2. Testing retry mechanism on initially broken output...')
broken_first_try = '{"name": "Incomplete json'
def dummy_llm_repair(prompt, system=None):
    return '{"name": "Repaired Candidate", "skills": ["python"]}'

recovered = extract_json(broken_first_try, ProfileJSON, max_retries=1, generate_fn=dummy_llm_repair)
print(f'Successfully recovered and extracted: {recovered.name} with {recovered.skills}')
````
